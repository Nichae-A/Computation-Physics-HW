```python
from numpy import exp

# part a: solve 5*e^-x + x - 5 = 0 using binary search (bisection)

accuracy = 10 ** -6

def find_root(f, x1, x2, accuracy):

#Finds a root of f(x) = 0 in the interval (x1, x2) using bisection.
#Requires f(x1) and f(x2) to have opposite signs.
#x1: lower bound of interval
#x2: upper bound of interval

    def midpoint(x, y):
        return (x + y) / 2

    def have_same_sign(x, y):
        if (x < 0 and y < 0) or (x > 0 and y > 0):
            return True
        else:
            return False

    f1, f2 = f(x1), f(x2)
    if have_same_sign(f1, f2):
        raise ValueError("f(x1) and f(x2) must have opposite signs "
                          "-- the interval does not have a root.")

    while abs(x1 - x2) > accuracy:
        x = midpoint(x1, x2)
        fx = f(x)
        if have_same_sign(f1, fx):
            x1, f1 = x, fx
        elif have_same_sign(fx, f2):
            x2, f2 = x, fx
        else:
            # fx is (numerically) exactly zero so we've hit the root
            return x

    return midpoint(x1, x2)

def f(x):
    return 5 * exp(-x) + x - 5

root = find_root(f, 4, 6, accuracy)
print("root =", root)
print("f(root) =", f(root))

#Part (b): solve 5*e^-x + x - 5 = 0 via bisection, then find b

accuracy = 10 ** -6

def find_root(f, x1, x2, accuracy):
#Bisection: find a root of f(x)=0 in (x1,x2), assuming f(x1),f(x2) have opposite signs

    def midpoint(x, y):
        return (x + y) / 2

    def have_same_sign(x, y):
        return (x < 0 and y < 0) or (x > 0 and y > 0)

    f1, f2 = f(x1), f(x2)
    if have_same_sign(f1, f2):
        raise ValueError("interval does not have a root")

    while abs(x1 - x2) > accuracy:
        x = midpoint(x1, x2)
        fx = f(x)
        if have_same_sign(f1, fx):
            x1, f1 = x, fx
        elif have_same_sign(fx, f2):
            x2, f2 = x, fx
        else:
            return x

    return midpoint(x1, x2)

def f(x):
    return 5 * exp(-x) + x - 5

x_root = find_root(f, 4, 6, accuracy)
print(f"Part (b): x = {x_root:.6f}   (f(x) = {f(x_root):.3e})")

# physical constants (SI units)
h  = 6.62607015e-34   # Planck's constant, J*s
c  = 299792458.0      # speed of light, m/s
kB = 1.380649e-23     # Boltzmann's constant, J/K

b = h * c / (kB * x_root)
print(f"Wien displacement constant b = {b:.6e} m*K")
print(f"                           b = {b*1e3:.6f} mm*K")

#Part (c): Sun's surface temperature from peak wavelength

lam_sun = 502e-9   
T_sun = b / lam_sun
print(f"\nPart (c): Sun's peak wavelength = {lam_sun*1e9:.0f} nm")
print(f"Estimated surface temperature T = {T_sun:.1f} K")
```
